````r
# Mean of the last w weeks (this week and the w - 1 before it)
# .complete = TRUE gives NA for the first w - 1 weeks, which have too few weeks
roll_mean = function(x, w) slider::slide_dbl(x, mean, .before = w - 1, .complete = TRUE)

lag_data = dengue |>
    select(
        city, year, weekofyear, week_start_date,
        humidity = reanalysis_specific_humidity_g_per_kg,
        dew_point = reanalysis_dew_point_temp_k,
        min_temp = station_min_temp_c,
        air_temp = reanalysis_air_temp_k,
        precip = precipitation_amt_mm
    ) |>
    left_join(labels, by = c('city', 'year', 'weekofyear')) |>
    arrange(city, week_start_date) |>
    group_by(city) |>
    fill(everything(), .direction = 'down') |>
    mutate(
        across(
            c(humidity, dew_point, min_temp, air_temp, precip),
            list(
                ma2 = ~roll_mean(.x, 2),
                ma4 = ~roll_mean(.x, 4),
                ma6 = ~roll_mean(.x, 6),
                ma8 = ~roll_mean(.x, 8)
            ),
            .names = '{.col}_{.fn}'
        ),
        cases_lag1 = lag(total_cases, 1),
        cases_lag2 = lag(total_cases, 2),
        cases_lag4 = lag(total_cases, 4),
        cases_ma4 = roll_mean(lag(total_cases, 1), 4),
        cases_ma8 = roll_mean(lag(total_cases, 1), 8),
        week_sin = sin(2 * pi * weekofyear / 52),
        week_cos = cos(2 * pi * weekofyear / 52)
    ) |>
    ungroup() |>
    drop_na() |>
    # Keep only what the model uses: the rolling means, case lags and season
    select(-c(year, week_start_date, humidity, dew_point, min_temp, air_temp, precip))
````

## Interpreting the lag-feature results

The lag model feeds in the case counts observed **up to `t-1`** (`cases_lag1`, `cases_lag2`, `cases_lag4`, and rolling means of past cases) together with the weather for the **same week** as the target. This makes the test evaluation a **rolling one-step-ahead** exercise: each week is scored using that week's *actual* past cases, then we move on to the next week. It is **not** a fixed-origin multi-week forecast, where you would predict weeks 1..H ahead from a single cut-off using only data known at that cut-off.

Likewise, the weather features are contemporaneous with the target week, so the results assume the target week's weather is already available at prediction time (a nowcast). A genuine multi-week-ahead system would also need to forecast the weather and would have to propagate its own predicted cases forward instead of using observed lags.

**AI-assisted interpretation.** In the San Juan lag-feature model the lagged case-count features dominate by a wide margin: `cases_lag1` (ranger impurity importance ≈ 5.4e5), `cases_lag2` (≈ 4.4e5), `cases_ma4` (≈ 3.9e5), `cases_ma8` (≈ 2.4e5) and `cases_lag4` (≈ 1.8e5) are the top five, roughly an order of magnitude above every weather feature. In other words, recent observed case counts — not the weather — carry most of the predictive signal, which matches the large MAE drop (San Juan 21.5 → 10.0, Iquitos 6.63 → 5.27).

Among the weather features the *longer* 8-week rolling means of `min_temp` (≈ 3.1e4), `humidity` (≈ 3.1e4) and `dew_point` (≈ 2.0e4) matter most, with the seasonal terms (`week_sin` ≈ 2.7e4, `weekofyear` ≈ 2.6e4) ranking just below them; the short 2-week weather means and precipitation (`precip_ma2` ≈ 1.5e3, the weakest) add little. This is consistent with dengue's slow seasonal build-up rather than week-to-week weather shocks. Caveat: these are ranger impurity importances computed on the San Juan training data (biased toward continuous/high-cardinality features and not a causal statement), and they are reported for San Juan only.
